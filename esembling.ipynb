```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Ai Medicaps(24 08 26)/ensemble ML/Clean_dataset.csv")
```

```python
df
```

```python
x = df.drop(columns = ['Price_inr'])
```

```python
y = df['Price_inr']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2 ,  random_state = 42 )
```

```python
#ml flow n
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
```

```python
cat_fea = ['Company','TypeName','Cpu','OpSys','Gpu']
```

```python
step1 = ColumnTransformer(transformers = [('col_tnf', OneHotEncoder(sparse_output = False, handle_unknown ='ignore'),cat_fea)], remainder = 'passthrough')
step2 = LinearRegression()
```

```python
pipe = Pipeline ([('step1',step1),('step2',step2)])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred = pipe.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
mse = mean_squared_error(y_test, y_pred)
mse
```

```python
r2 = r2_score(y_test, y_pred)*100
r2
```

```python
men = mean_absolute_error(y_test, y_pred)
men
```

```python
#GradientBoostingRegressor
step1 = ColumnTransformer(transformers = [
    ('col_tnf',OneHotEncoder(sparse_output = False, handle_unknown = 'ignore'),cat_fea)
],remainder = 'passthrough')

step2 = GradientBoostingRegressor( n_estimators=500)

pipe = Pipeline([('step1',step1),('step2',step2)])

#Training the model stored in Pipe
pipe.fit(x_train, y_train)

y_pred = pipe.predict(x_test)  #Storing the Pred made by the machine
print("R_2 Score for GradientBoostingRegressor",r2_score(y_test,y_pred))
print("Mean Absolute Error GradientBoostingRegressor",mean_absolute_error(y_test,y_pred))
print("Mean Squared Error GradientBoostingRegressor",mean_squared_error(y_test,y_pred))
```
